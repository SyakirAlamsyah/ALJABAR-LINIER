## 1️⃣ Python vs Traditional Programming

Think of traditional compiled languages (like C or Java) as **writing a full letter, sealing the envelope, and mailing it** — you only find out if there's a mistake after it's sent and opened (compiled and run).

Python is more like **having a live conversation** — you say something (write a line of code), and you get a response immediately. This is because Python is an **interpreted language**: it runs your code line-by-line, instantly.

Let's see this "instant feedback" in action below. 👇


````python
# Python evaluates and shows results immediately - no separate "compile" step needed
5 * 8 + 2
````

## 4️⃣ Syntax, Indentation, and `print()` 🖨️

Python has a few golden rules:

1. **`print()`** displays output to the screen — it's how Python "talks back" to you.
2. **Indentation matters!** Unlike many languages that use `{ }` to group code, Python uses **whitespace (indentation)** to define blocks of code (like inside an `if` statement or a loop). Incorrect indentation will cause an error.
3. Statements are usually written **one per line**, and don't need semicolons.

Let's practice step by step. 👇


````python
# A basic print statement
print("Hello, Python!")
````

````python
print("Hello, Fajri")
````

````python
# print() can take multiple arguments, separated by commas
# Python automatically adds a space between them
print("My name is", "Alex", "and I am learning Python!")
````

### ⚠️ What NOT To Do: Broken Indentation

Below is an example of **incorrect indentation** (shown as text, not run, so it doesn't break our notebook). In Python, this would raise an `IndentationError`:

```python
if 5 > 2:
print("Five is greater than two")   # ❌ Error! This line needs to be indented
```

Python expects the code **inside** the `if` block to be indented consistently (usually 4 spaces). Let's see the **correct** version in the next cell. ✅


````python
# Correct indentation - the line inside the "if" block is indented with 4 spaces
if 5 > 2:
  print("Five is greater than two")  # ✅ This works!
````

## 5️⃣ Comments & Code Readability 💬

Comments start with `#` and are ignored by Python when the code runs. They're notes **for humans**, not the computer. Good comments explain *why* something is done, not just *what* (the code already shows *what*).


````python
# ❌ BAD comment: just repeats what the code obviously does
x = 5  # set x to 5

# ✅ GOOD comment: explains the purpose/reasoning
x = 5  # number of students in the study group, used later to split tasks evenly

print(x)
````
